# Rozpoznawanie ręcznie pisanych cyfr

**Projekt w Pythonie z użyciem TensorFlow i Tkinter**

Krótka prezentacja projektu: założenia, technologie, architektura i sposób działania.

## Cel projektu

Celem było stworzenie prostej aplikacji, która:

- rozpoznaje cyfry od **0 do 9**
- pozwala użytkownikowi narysować cyfrę na ekranie
- zwraca przewidywanie modelu oraz poziom pewności
- działa lokalnie jako aplikacja desktopowa

## Założenia projektu

- model ma uczyć się na popularnych zbiorach cyfr
- prosty i czytelny interface
- zapisany model ma się wczytywać przy kolejnym uruchomieniu
- projekt jest łatwy do uruchomienia i przetestowania

## Technologie

- **Python** — główny język projektu
- **TensorFlow / Keras** — budowa i trening modelu
- **TensorFlow Datasets** — pobranie EMNIST
- **NumPy** — operacje na tablicach i danych
- **Pillow** — praca z obrazem i preprocessing
- **Tkinter** — prosty interfejs graficzny

## Architektura projektu

Projekt składa się z czterech głównych części:

- **data.py** — wczytywanie i przygotowanie danych
- **model.py** — budowa, trening i zapis modelu
- **gui.py** — interfejs graficzny i obsługa rysowania
- **main.py** — uruchomienie całej aplikacji

Schemat działania:

`dane → trening modelu → zapis modelu → GUI → rysunek użytkownika → predykcja`

## Jak działają dane

Do treningu używane są dwa zbiory:

- **MNIST**
- **EMNIST digits**

Dane są:

- normalizowane do zakresu **0–1**
- dopasowane do formatu **28 × 28 × 1**
- w EMNIST dodatkowo korygowane, bo obrazy są obrócone
- łączone w jeden zbiór treningowy i testowy

## Model

Użyty model to mała **sieć konwolucyjna CNN**.

W skrócie:

- warstwy **Conv2D** wykrywają cechy obrazu
- **MaxPooling** zmniejsza rozmiar danych
- warstwa **Dense** podejmuje decyzję końcową
- ostatnia warstwa **Softmax** zwraca prawdopodobieństwa dla cyfr 0–9

## Trening i zapis modelu

- przy pierwszym uruchomieniu model może zostać wytrenowany automatycznie
- po treningu zapisuje się do pliku `digit_model.keras`
- przy kolejnych uruchomieniach model jest po prostu wczytywany
- dzięki temu aplikacja startuje szybciej

## Ograniczenia i możliwe ulepszenia

**Ograniczenia:**

- działa tylko dla cyfr, nie dla liter
- jakość predykcji zależy od sposobu rysowania
- na Windows TensorFlow zwykle działa bez GPU

**Możliwe ulepszenia:**

- lepszy preprocessing obrazu
- rozbudowany interfejs
- obsługa liter z pełnego EMNIST
- wizualizacja architektury modelu

## Podsumowanie

Projekt pokazuje pełny prosty pipeline ML:

- przygotowanie danych
- trening modelu
- zapis i ponowne użycie
- aplikacja graficzna dla użytkownika

To krótki, ale kompletny przykład połączenia **uczenia maszynowego** z **aplikacją desktopową**.
